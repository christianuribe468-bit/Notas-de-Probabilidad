#$Ejercicio$ 1
##Al lanzar dos dados, calcula la probabilidad de que la suma de sus caras sea exactamente 5.

```python
casos_posibles = 36
casos_favorables = 0

for i in range(1, 7):
    for j in range(1, 7):
        if i + j == 5:
            casos_favorables += 1

probabilidad = casos_favorables / casos_posibles

print("Casos posibles:", casos_posibles)
print("Casos favorables:", casos_favorables)
print("Probabilidad:", probabilidad)
print("Porcentaje:", probabilidad * 100, "%")

```

#$Ejercicio$ 2

##En una contraseña de 3 dígitos, del 0 al 9, calcula la probabilidad de que ninguno de los dígitos se repita.

```python
casos_posibles = 10 * 10 * 10
casos_favorables = 10 * 9 * 8

probabilidad = casos_favorables / casos_posibles

print("Casos posibles:", casos_posibles)
print("Casos favorables:", casos_favorables)
print("Probabilidad:", probabilidad)
print("Porcentaje:", probabilidad * 100, "%")
```

Tienes 8 canciones distintas y las pones en un orden aleatorio, con todos los órdenes igualmente probables. Calcula la probabilidad de que tu favorita quede primera y la que menos te guste quede última.

#$Ejercicio$ $4$

Se lanzan tres dados equilibrados. ¿Qué es más probable, que la suma sea 9 o que la suma sea 10? ¿Por cuántos resultados le gana una suma a la otra?

```python
casos_posibles = 6 * 6 * 6
suma_9 = 0
suma_10 = 0

for i in range(1, 7):
    for j in range(1, 7):
        for k in range(1, 7):
            if i + j + k == 9:
                suma_9 += 1
            if i + j + k == 10:
                suma_10 += 1

print("Casos posibles:", casos_posibles)
print("Resultados con suma 9:", suma_9)
print("Resultados con suma 10:", suma_10)

print("Probabilidad de suma 9:", suma_9 / casos_posibles)
print("Probabilidad de suma 10:", suma_10 / casos_posibles)

print("Diferencia de resultados:", abs(suma_9 - suma_10))
```
