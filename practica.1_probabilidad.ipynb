Por ejemplo, en matemáticas, si evaluamos la operación 2+3, el resultado siempre será 5. De manera similar, si calculamos el área de un círculo usando la fórmula

Un experimento aleatorio es aquel que, aún cuando se repita bajo las mismas condiciones, no produce siempre el mismo resultado y no es posible predecir con certeza cuál ocurrirá.

LANZAMIENTO DE MONEDA

```python
import random

Dado_lados = ["1", "2", "3", "4", "5", "6"]

def lanzar_dado():
    return random.choice(Dado_lados)

lanzar_dado()
```
